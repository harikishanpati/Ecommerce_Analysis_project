```python

import pandas as pd


import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import plotly.colors as colors

pio.templates.default = 'plotly_white'
```

```python
data = pd.read_csv('Sample - Superstore.csv', encoding = 'latin1')
```

```python
data.head()
```

```python
data.describe()
```

```python
data.info()
```

```python
data['Order Date'] = pd.to_datetime(data['Order Date'])
```

```python
data.info()
```

```python
data['Ship Date'] = pd.to_datetime(data['Ship Date'])
```

```python
data.head()
```

```python
data['Order month'] = data['Order Date'].dt.month
data['Order year'] = data['Order Date'].dt.year
data['Order of the week'] = data['Order Date'].dt.dayofweek
```

```python
data.head()
```

#monthly sales 

```python
sales_by_month = data.groupby('Order month')['Sales'].sum().reset_index()
print(sales_by_month)
```

```python
sales_by_month
```

```python
fig = px.line(sales_by_month, x = 'Order month', y = 'Sales',
              title = 'Monthly_data_analysis')
fig.show()
```

```python
sales_by_category = data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
fig = px.pie(sales_by_category,
              values = 'Sales',
              names = 'Category',
              hole = 0.1,
            )
fig.update_traces(textposition = 'inside', textinfo = 'percent+label')
fig.update_layout(title_text = 'sales by category', title_font = dict(size=26))

fig.show()

```

```python
sales_by_subcategory = data.groupby('Sub-Category')['Sales'].sum().reset_index()
```

```python
sales_by_subcategory
```

```python
fig = px.bar(sales_by_subcategory, x = 'Sub-Category', y = 'Sales', title = 'sales by sub-category')
fig.show()
```

```python
profit_month = data.groupby('Order month')['Profit'].sum().reset_index()
```

```python
profit_month
```

```python
fig = px.bar(profit_month, x = 'Order month', y = 'Profit', title = 'profit of the month')

fig.show()
```

```python
profit_by_category = data.groupby('Category')['Profit'].sum().reset_index()
```

```python
profit_by_category
```

```python
fig = px.pie(profit_by_category, values = 'Profit',
             names = 'Category', hole = 0.5, title = 'profit category analysis')

fig.show()
             
```
